```python
import os

def verify_chunks(output_chunks, data_dir):
    print("\n--- BÁO CÁO KIỂM TRA CHUNKING ---")
    
    processed_files = set(chunk['metadata']['source_file'] for chunk in output_chunks)
    print(f"Tổng số file đã xử lý: {len(processed_files)}")

    if processed_files:
        sample_file = list(processed_files)[0]

        sample_chunks = [c for c in output_chunks if c['metadata']['source_file'] == sample_file]
        sample_chunks.sort(key=lambda x: x['metadata']['chunk_index'])

        total_chunk_chars = sum(len(c['text_content']) for c in sample_chunks)

        full_file_path = os.path.join(data_dir, sample_file)
        
        try:
            with open(full_file_path, 'r', encoding='utf-8') as f:
                original_content = f.read()
            total_original_chars = len(original_content)
            
            print(f"\nKiểm tra file mẫu: {sample_file}")
            print(f"- Số lượng chunk tạo ra: {len(sample_chunks)}")
            print(f"- Tổng số ký tự file gốc: {total_original_chars}")
            print(f"- Tổng số ký tự sau khi chunk (có overlap): {total_chunk_chars}")
            
            if total_chunk_chars >= total_original_chars:
                print("=> Trạng thái: Đạt chuẩn (Dữ liệu đầy đủ, không bị mất mát).")
            else:
                print("=> Trạng thái: CẢNH BÁO! Số ký tự chunk ít hơn file gốc. Có thể bị mất dữ liệu.")
        except FileNotFoundError:
            print(f"\n[Lỗi] Không tìm thấy file gốc tại đường dẫn: {full_file_path}")
```

```python
output = r'.\all_chunks.json'
dir = r'MockProject_062026_NhomAI\data_clean\RAG'
```

```python
import json

with open(output, 'r', encoding='utf-8') as f:
    chunks_data = json.load(f)

verify_chunks(chunks_data, dir)
```
